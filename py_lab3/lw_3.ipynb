В нашем магазине все товарные коды всегда имеют формат "B123", где "B" — это буква, а "123" — цифры. Напишите программу, которая проверяет, соответствует ли товарный код этому формату.

Нам часто поступает список книг в виде строки, где данные о каждой книге разделены пробелами или запятыми. Напишите программу, которая будет извлекать информацию о каждой книге (название, автор, год издания) и выводить их по категориям.

````python
def extract_book_info(data):
    items = data.split(",")

    titles = []
    authors = []
    years = []

    for i in range(0, len(items), 3):
        titles.append(items[i])
        authors.append(items[i + 1])
        years.append(items[i + 2])

    return titles, authors, years


data = "The Great Gatsby,F. Scott Fitzgerald,1925,1984,George Orwell,1949,To Kill a Mockingbird,Harper Lee,1960"

titles, authors, years = extract_book_info(data)

print("Titles:", titles)
print("Authors:", authors)
print("Years:", years)
````

Нам необходимо собирать контактные данные клиентов.
Напишите программу, которая проверяет, соответствует ли введенный адрес электронной почты клиента **правильному** формату.

Правильный формат адреса электронной почты обычно определяется следующими правилами:

1. **Имя пользователя (local part)**:
   - Может содержать буквы, цифры, точки (`.`), дефисы (`-`), знаки подчеркивания (`_`).
   - Не может начинаться или заканчиваться точкой (`.`).
   - Не допускается последовательность точек (`..`).

2. **Символ `@`**:
   - Разделяет имя пользователя и домен.
   - Должен присутствовать ровно один раз.

3. **Доменная часть (domain)**:
   - Состоит из букв, цифр, дефисов (`-`) и точек (`.`).
   - Должен содержать хотя бы одну точку.
   - Доменное имя не может начинаться или заканчиваться дефисом (`-`) или точкой (`.`).

4. **Домен верхнего уровня (TLD)**:
   - Обычно состоит из 2 или более букв (например, `.com`, `.org`, `.ru`).



#### Примеры ввода и проверки:
- **Допустимые адреса**:
  - `example@gmail.com`
  - `user.name@sub.domain.org`
  - `first_last123@domain.co.uk`

- **Недопустимые адреса**:
  - `user@.com` (домен не может начинаться с точки)
  - `@domain.com` (нет имени пользователя)
  - `user@domain` (нет TLD)
  - `user@@domain.com` (две `@`)

````python
import re

def check_email_format(email):
    # Регулярка учитывает:
    # - имя пользователя: буквы, цифры, точки, дефисы, подчеркивания
    #   (не начинается и не заканчивается точкой, нет двойных точек)
    # - ровно один @
    # - домен: буквы, цифры, дефисы, точки, хотя бы одна точка
    # - TLD: минимум 2 буквы
    pattern = r'^[A-Za-z0-9]+([._-][A-Za-z0-9]+)*@[A-Za-z0-9-]+(\.[A-Za-z0-9-]+)*\.[A-Za-z]{2,}$'

    if re.match(pattern, email):
        return True
    else:
        return False

# Примеры использования
emails = [
    "example@gmail.com",
    "user.name@sub.domain.org",
    "first_last123@domain.co.uk",
    "user@.com",
    "@domain.com",
    "user@domain",
    "user@@domain.com"
]

for e in emails:
    print(e, "->", check_email_format(e))
````

````python
def check_restrictions(data):
    text = data.lower()
    restrictions = ["глютен", "лактоза", "орехи"]
    for r in restrictions:
        if r in text:
            return True
    return False

client_data = "В моей диете исключены продукты с лактозой и глютеном."
if check_restrictions(client_data):
    print("Обнаружены пищевые ограничения.")
else:
    print("Пищевых ограничений не обнаружено.")
````

Предположим, у вас есть текст, который может содержать несколько телефонных номеров. Каждый номер телефона записан в формате:

- `(XXX) XXX-XXXX`
- `XXX-XXX-XXXX`
- `XXX.XXX.XXXX`
- `XXX XXX XXXX`

Где `X` — цифра от 0 до 9.

Напишите программу, которая будет искать все телефонные номера в тексте и выводить их.


````python
import re

def extract_phone_numbers(text):
    # Паттерн для поиска телефонов:
    # - (XXX) XXX-XXXX
    # - XXX-XXX-XXXX
    # - XXX.XXX.XXXX
    # - XXX XXX XXXX
    pattern = r'(\(\d{3}\)\s\d{3}-\d{4})|(\d{3}[-. ]\d{3}[-. ]\d{4})'

    # findall возвращает список кортежей, поэтому нужно вытащить непустые элементы
    matches = re.findall(pattern, text)

    # Собираем результат в один список
    result = [m[0] if m[0] else m[1] for m in matches]
    return result


text = "Hi, here are some phone numbers: (123) 456-7890, 987-654-3210, 555.555.5555, 111 222 3333."
print(extract_phone_numbers(text))
````

У тебя есть лог-файл веб-сервера в формате:

```
[2026-02-18 14:23:01] INFO user_id=U-12345 ip=192.168.1.10 action=LOGIN
[2026-02-18 14:25:44] ERROR user_id=U-999 ip=10.0.0.256 action=PAYMENT
[2026-02-18 14:26:12] INFO user_id=U-54321 ip=172.16.0.1 action=LOGOUT
```

1. Напишите функцию:

```python
def parse_logs(text: str) -> list[dict]:
```

2. С помощью regex:

   * Извлечь `timestamp`, `level`, `user_id`, `ip`, `action`
   * Пропустить строки с **невалидным IPv4**
   * Проверить:

     * `user_id` формат: `U-` + 3–6 цифр
     * `action` только: LOGIN, LOGOUT, PAYMENT

3. Верните список словарей только с корректными записями.

````python
# Напиши решение тут
import ipaddress

def parse_logs(text: str) -> list[dict]:
    valid_entries = []

    # Паттерн для извлечения полей
    pattern = re.compile(
        r'\[(?P<timestamp>[\d-]+\s[\d:]+)\]\s+'        # timestamp
        r'(?P<level>\w+)\s+'                            # level
        r'user_id=(?P<user_id>U-\d{3,6})\s+'           # user_id
        r'ip=(?P<ip>[0-9.]+)\s+'                        # ip (проверим валидность отдельно)
        r'action=(?P<action>LOGIN|LOGOUT|PAYMENT)'      # action
    )

    # Проходим по каждой строке лога
    for line in text.splitlines(): #splitlines разбивает на список строк
        match = pattern.search(line)
        if match:
            entry = match.groupdict()

            # Проверяем валидность IPv4
            try:
                ipaddress.IPv4Address(entry['ip'])
            except ipaddress.AddressValueError:
                continue  # пропускаем невалидный IP

            # Если всё ок — добавляем запись
            valid_entries.append(entry)

    return valid_entries


log_text = """[2026-02-18 14:23:01] INFO user_id=U-12345 ip=192.168.1.10 action=LOGIN
[2026-02-18 14:25:44] ERROR user_id=U-999 ip=10.0.0.256 action=PAYMENT
[2026-02-18 14:26:12] INFO user_id=U-54321 ip=172.16.0.1 action=LOGOUT"""

result = parse_logs(log_text)
for entry in result:
    print(entry)
````

Дан текст:

```html
<div class="container">
  <div class="item" data-id="123">
      <span>Email: test@example.com</span>
  </div>
  <div class="item" data-id="456">
      <span>Email: admin@mail.org</span>
  </div>
</div>
```

Напишите функцию:

```python
def extract_emails_from_items(html: str) -> list[tuple[str, str]]:
```

Функция должна:

* Найти только `<div class="item" data-id="...">`
* Извлечь:

  * `data-id`
  * email внутри `<span>`

Верните список:

```python
[("123", "test@example.com"), ("456", "admin@mail.org")]
```


````python
import re

def extract_emails_from_items(html: str) -> list[tuple[str, str]]:
    pattern = re.compile(
        r'<div\s+class="item"\s+data-id="(?P<data_id>\d+)">.*?'
        r'<span>.*?([A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,})</span>',
        re.DOTALL
    )

    # Найти все совпадения
    matches = pattern.findall(html)

    # Каждое совпадение — кортеж (data_id, email)
    result = [(m[0], m[1]) for m in matches]
    return result



html_text = """<div class="container">
  <div class="item" data-id="123">
      <span>Email: test@example.com</span>
  </div>
  <div class="item" data-id="456">
      <span>Email: admin@mail.org</span>
  </div>
</div>"""

print(extract_emails_from_items(html_text))
````


Создайте функцию валидирующую пароли:

```python
def is_valid_password(password: str) -> bool:
```

Пароль валиден, если:

* 10–20 символов
* хотя бы:

  * 2 заглавные буквы
  * 2 цифры
  * 1 спецсимвол из `@#$%^&+=`
* не содержит:

  * 3 одинаковых символа подряд
  * пробелов
* не начинается и не заканчивается цифрой

````python
import re

def is_valid_password(password: str) -> bool:
    if not 10 <= len(password) <= 20:
        return False

    if ' ' in password:
        return False

    if password[0].isdigit() or password[-1].isdigit():
        return False

    if len(re.findall(r'[A-Z]', password)) < 2:
        return False

    if len(re.findall(r'\d', password)) < 2:
        return False

    if len(re.findall(r'[@#$%^&+=]', password)) < 1:
        return False

    if re.search(r'(.)\1\1', password):
        return False

    return True


# Примеры
print(is_valid_password("Ab12@DefGh"))
print(is_valid_password("Ab1@DefGh"))
print(is_valid_password("ABC12@@defgh"))
print(is_valid_password("Ab111@DefGh"))
print(is_valid_password("1Ab12@DefGh2"))
````

Дан текст с несколькими SQL-запросами:

```sql
SELECT u.id, u.name, o.amount
FROM users u
JOIN orders o ON u.id = o.user_id
WHERE o.amount > 100;

SELECT * FROM products;
```

Напишите функцию:

```python
def extract_sql_structure(query: str) -> list[dict]:
```

Функция должна для каждого SELECT-запроса извлечь:

* список полей
* таблицы после FROM
* таблицы после JOIN (если есть)
* условие WHERE (если есть)

Верните структуру:

```python
[
  {
    "fields": ["u.id", "u.name", "o.amount"],
    "from": "users",
    "joins": ["orders"],
    "where": "o.amount > 100"
  },
  ...
]
```

````python
qimport re
# функция принимает строку с SQL запросом
def extract_sql_structure(query: str) -> list[dict]:
    # Разделяем несколько SELECT-запросов по ;
    queries = [q.strip() for q in query.split(';') if q.strip()]
    #q.strip()
    #Убирает лишние пробелы и переносы строк в начале и конце каждой строки
    #if q.strip()
    #Исключает пустые строки, которые могут появиться после split(';')
    results = []
    #проходим по каждому SQL запросу q-это один SQL-запрос без лишних пробелов
    for q in queries:
        q_lower = q.lower()  # для поиска ключевых слов без учёта регистра
        # Поля: между SELECT и FROM
        fields_match = re.search(r'select\s+(.*?)\s+from\s', q_lower, re.IGNORECASE | re.DOTALL)
        #re.search ищём шаблон
        #select\s+ → слово SELECT + один или несколько пробелов
        #(.*?) → все символы (поля) до FROM
        fields = []
        if fields_match:
            # Берём оригинальный текст из исходного запроса
            start, end = fields_match.span(1) #<- Возвращает индексы начала и конца совпадения группы (.*?)
                        #Используем эти индексы, чтобы взять оригинальный текст из запроса, а не из q_lower
            fields_text = q[start:end].strip()
            fields = [f.strip() for f in fields_text.split(',')]

        # FROM таблица
        from_match = re.search(r'from\s+([^\s]+)', q_lower, re.IGNORECASE)
        from_table = from_match.group(1) if from_match else None
        #from\s+([^\s]+) → ищем FROM + пробелы + текущее слово
        #[^\s]+ → любые символы до следующего пробела

        # JOIN таблицы (может быть несколько)
        joins = re.findall(r'join\s+([^\s]+)', q_lower, re.IGNORECASE)
        #re.findall возвращает все совпадения, потому что JOIN может быть несколько
        #[^\s]+ → имя таблицы после JOIN

        # WHERE условие (берём до конца запроса или до следующего ключевого слова)
        where_match = re.search(r'where\s+(.*)', q_lower, re.IGNORECASE | re.DOTALL)
        where_clause = None
        if where_match:
            start, end = where_match.span(1)
            # Берём текст из исходного запроса
            where_clause = q[start:end].strip()
            # Убираем возможные лишние переносы строк и точки с запятой
            where_clause = where_clause.rstrip(';').replace('\n', ' ').strip()
            #where\s+(.*) → всё после WHERE до конца строки
            #q[start:end] → берём оригинальный текст для точности
            #rstrip(';') → убираем точку с запятой
            #replace('\n', ' ') → перенос строки превращаем в пробел
            #strip() → убираем лишние пробелы

        results.append({
            "fields": fields,
            "from": from_table,
            "joins": joins,
            "where": where_clause
        })

    return results

# Пример
sql_text = """
SELECT u.id, u.name, o.amount
FROM users u
JOIN orders o ON u.id = o.user_id
WHERE o.amount > 100;

SELECT * FROM products;
"""

import pprint
pprint.pprint(extract_sql_structure(sql_text))
````
